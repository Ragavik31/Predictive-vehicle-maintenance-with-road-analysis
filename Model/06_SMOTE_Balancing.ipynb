```python
import pandas as pd
import joblib
from collections import Counter

from imblearn.over_sampling import SMOTE
```

```python
import pandas as pd

# Load the train and test datasets saved in Notebook 5
train_df = pd.read_csv("data/processed/train_dataset.csv")
test_df = pd.read_csv("data/processed/test_dataset.csv")

# Separate features and target
target = "Maintenance_Required"

X_train = train_df.drop(columns=[target])
y_train = train_df[target]

X_test = test_df.drop(columns=[target])
y_test = test_df[target]

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)
```

```python
print("Before SMOTE")
print(Counter(y_train))
```

```python
print("After SMOTE")
print(Counter(y_train_smote))
```

```python
joblib.dump(X_train_smote,
            "data/processed/X_train_smote.pkl")

joblib.dump(y_train_smote,
            "data/processed/y_train_smote.pkl")
```

```python
comparison = pd.DataFrame({
    "Before SMOTE": Counter(y_train),
    "After SMOTE": Counter(y_train_smote)
})

comparison
```

```python
train_smote_df = X_train_smote.copy()

train_smote_df["Maintenance_Required"] = y_train_smote.values

train_smote_df.to_csv(
    "data/processed/train_dataset_smote.csv",
    index=False
)

print("SMOTE training dataset saved successfully.")
```
